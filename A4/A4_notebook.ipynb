# Time-series NDVI Analysis for Salzburg, Austria

This notebook demonstrates the usage of functions defined in `timeseries_ndvi.py` for analyzing NDVI data over time.

## Importing the Necessary Libraries and Module

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from timeseries_ndvi import calculate_ndvi, plot_ndvi_timeseries, analyze_ndvi_trends
```

## Simulated NDVI Data
For demonstration purposes, we'll create some simulated NDVI data for Salzburg, Austria.

```python
# Simulated dates and NDVI values
dates = pd.date_range(start='2020-01-01', periods=24, freq='M')
nir_values = np.random.uniform(0.3, 0.9, size=len(dates))
red_values = np.random.uniform(0.1, 0.5, size=len(dates))
ndvi_values = calculate_ndvi(nir_values, red_values)
```

```python
## Plotting NDVI Time-series
plot_ndvi_timeseries(dates, ndvi_values)
```

```python
## Analyzing NDVI Trends
trend_info = analyze_ndvi_trends(dates, ndvi_values)
print(f"Slope: {trend_info['slope']}")
print(f"Intercept: {trend_info['intercept']}")
print(f"R-squared: {trend_info['r_squared']}")
print(f"P-value: {trend_info['p_value']}")
print(f"Standard Error: {trend_info['std_err']}")
```
